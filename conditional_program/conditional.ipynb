```python
age = int(input("Enter your age "))
gender = input("Press M for Male and F for Female ")

# more than 2 conditions
if gender=='M' or gender=='m':
  if age>=21:
    print("You are eligible to vote")
  else:
    print("You are male but not eligible to vote because your age is below than 21")
elif gender=='F' or gender=='f':
  if age>=18:
    print("You are eligible to vote")
  else:
    print("You are female but not eligible to vote because your age is below than 18")
else:
    print("Please input M or F for gender")
```
